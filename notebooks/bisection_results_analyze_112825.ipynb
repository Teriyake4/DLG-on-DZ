```python
upper_thresh = 10

# Filter data and calculate lost observations
dlg_data = df['DLG MSE'][df['DLG MSE'] < upper_thresh]
idlg_data = df['iDLG MSE'][df['iDLG MSE'] < upper_thresh]

# Calculate observations lost
dlg_lost = len(df['DLG MSE']) - len(dlg_data)
data2_lost = len(df['iDLG MSE']) - len(data2_data)

# Create the plot
_ = plt.figure(figsize=(30,15))
_ = plt.hist(dlg_data, color='lime', label='DLG MSE', alpha=0.5, bins=10)
_ = plt.hist(data2_data, color='blue', label='iDLG MSE', alpha=0.5, bins=10)
plt.legend(fontsize=30)

# Add text showing lost observations inside the plot
plt.text(0.02, 0.98, f'DLG: {dlg_lost} obs excluded (≥{upper_thresh})', 
         transform=plt.gca().transAxes, fontsize=24, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='lime', alpha=0.3))

plt.text(0.02, 0.90, f'iDLG: {data2_lost} obs excluded (≥{upper_thresh})', 
         transform=plt.gca().transAxes, fontsize=24, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='blue', alpha=0.3))
plt.xlabel('MSE', fontsize=30)
plt.xlabel('Num Experiments', fontsize=30)
```
